```python
# import python library
import pandas as pd
import holoviews as hv
import hvplot.pandas
import osmnx as osm
```

```python
USASK_boundary = osm.geocode_to_gdf("R13754325", by_osmid=True)

USASK_boundary
```

```python
USASK_boundary.plot()
```

```python
# Make interactive map
USASK_map = USASK_boundary.hvplot(geo=True, tiles = "EsriImagery", alpha = 0.5)
```

```python
hv.save(USASK_map, 'USASK_map.html')

# Display the map
USASK_map

```
