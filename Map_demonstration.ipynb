```python
# import python library
import pandas as pd
import holoviews as hv
import hvplot.pandas
import osmnx as osm
```

```python
USASK_boudary = osm.geocode_to_gdf(
            "W155526740",
            by_osmid= True

)
USASK_boudary
```

```python
USASK_boudary.plot()
```

```python
# Make interactive map
USASK_map = USASK_boudary.hvplot(geo=True, tiles = "EsriImagery", alpha = 0.5)
```

```python
hv.save(USASK_map, 'USASK_map.html')

# Display the map
USASK_map

```
